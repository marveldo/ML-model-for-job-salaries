```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
from sklearn.base import BaseEstimator, TransformerMixin

class MultiLabelBinarizerTransformer(BaseEstimator, TransformerMixin):
    def __init__(self, sparse_output=False):
        self.sparse_output = sparse_output
        self.mlb = MultiLabelBinarizer(sparse_output=sparse_output)

    def fit(self, X, y=None):
        self.mlb.fit(X.iloc[:, 0] if hasattr(X, 'iloc') else X[:, 0])
        return self

    def transform(self, X):
        col = X.iloc[:, 0] if hasattr(X, 'iloc') else X[:, 0]
        
        return self.mlb.transform(col)

    def get_feature_names_out(self, input_features=None):
        return self.mlb.classes_.tolist()
```

```python
X_P_train
```

```python
preprocessing = ColumnTransformer([
    ('languages', MultiLabelBinarizerTransformer(), ['languages']),
    ('frameworks', MultiLabelBinarizerTransformer(), ['frameworks'])
], remainder='passthrough')
```

```python
pipeline = Pipeline([
    ('preprocessor', preprocessing),
    ('regressor', LinearRegression())
])
```

```python
X_P_train
```

```python
pipeline.fit(X_P_train, y_P_train)
```

```python
y_pipeline_predict = pipeline.predict(X_P_test)
```
